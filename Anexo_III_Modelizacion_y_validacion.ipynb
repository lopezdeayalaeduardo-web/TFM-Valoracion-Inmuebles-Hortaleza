## CÓDIGO A3.1

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split, RepeatedKFold, cross_val_score
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
```

## CÓDIGO A3.2

```python
df = pd.read_csv("dataset_modelo_hortaleza.csv")

print("Dimensiones del dataset:", df.shape)
print("\nNúmero de viviendas:", len(df))
print("\nVariables disponibles:")
print(df.columns.tolist())
```

## CÓDIGO A3.3

```python
variable_objetivo = "price"

variables_predictoras = [
    "size", "rooms", "bathrooms", "floor", "hasLift", "status",
    "propertyType", "typology", "exterior", "newDevelopment",
    "latitude", "longitude", "dist_metro", "dist_bus", "dist_school",
    "dist_hospital", "dist_health_center", "dist_supermarket",
    "dist_park", "dist_sports"
]

X = df[variables_predictoras].copy()
y = df[variable_objetivo].copy()

print("Dimensiones de X:", X.shape)
print("Dimensiones de y:", y.shape)
```

## CÓDIGO A3.4

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print("Training dataset:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTesting dataset:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

## CÓDIGO A3.5

```python
variables_categoricas = [
    "floor", "hasLift", "status", "propertyType", "typology", "exterior"
]

variables_numericas = [
    "size", "rooms", "bathrooms", "newDevelopment", "latitude", "longitude",
    "dist_metro", "dist_bus", "dist_school", "dist_hospital",
    "dist_health_center", "dist_supermarket", "dist_park", "dist_sports"
]
```

## CÓDIGO A3.6

```python
preprocesador = ColumnTransformer(
    transformers=[
        ("categoricas",
         OneHotEncoder(handle_unknown="ignore", sparse_output=False),
         variables_categoricas)
    ],
    remainder="passthrough"
)

print("Preprocesador configurado correctamente.")
```

## CÓDIGO A3.7

```python
modelo_regresion_lineal = Pipeline(
    steps=[
        ("preprocesamiento", preprocesador),
        ("modelo", LinearRegression())
    ]
)

print("Modelo de Regresión Lineal configurado correctamente.")
```

## CÓDIGO A3.8

```python
modelo_regresion_lineal.fit(X_train, y_train)

print("Modelo de Regresión Lineal entrenado correctamente.")
```

## CÓDIGO A3.9

```python
y_pred_regresion_lineal = modelo_regresion_lineal.predict(X_test)

print("Predicciones realizadas correctamente.")
print("\nNúmero de predicciones:", len(y_pred_regresion_lineal))
```

## CÓDIGO A3.10

```python
mae_rl = mean_absolute_error(y_test, y_pred_regresion_lineal)
rmse_rl = np.sqrt(mean_squared_error(y_test, y_pred_regresion_lineal))
r2_rl = r2_score(y_test, y_pred_regresion_lineal)

print("Resultados de la Regresión Lineal")
print(f"MAE:  {mae_rl:,.2f} €")
print(f"RMSE: {rmse_rl:,.2f} €")
print(f"R²:   {r2_rl:.4f}")
```

## CÓDIGO A3.11

```python
resultados_modelos = pd.DataFrame({
    "Modelo": ["Regresión Lineal"],
    "MAE (€)": [mae_rl],
    "RMSE (€)": [rmse_rl],
    "R²": [r2_rl]
})

display(resultados_modelos)
```

## CÓDIGO A3.12

```python
modelo_random_forest = Pipeline(
    steps=[
        ("preprocesamiento", preprocesador),
        ("modelo", RandomForestRegressor(
            n_estimators=500,
            random_state=42,
            n_jobs=-1
        ))
    ]
)

print("Modelo Random Forest configurado correctamente.")
```

## CÓDIGO A3.13

```python
modelo_random_forest.fit(X_train, y_train)

print("Modelo Random Forest entrenado correctamente.")
```

## CÓDIGO A3.14

```python
y_pred_random_forest = modelo_random_forest.predict(X_test)

print("Predicciones realizadas correctamente.")
```

## CÓDIGO A3.15

```python
mae_rf = mean_absolute_error(y_test, y_pred_random_forest)
rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_random_forest))
r2_rf = r2_score(y_test, y_pred_random_forest)

print("Resultados de Random Forest")
print(f"MAE:  {mae_rf:,.2f} €")
print(f"RMSE: {rmse_rf:,.2f} €")
print(f"R²:   {r2_rf:.4f}")
```

## CÓDIGO A3.16

```python
resultados_modelos = pd.DataFrame({
    "Modelo": ["Regresión Lineal", "Random Forest"],
    "MAE (€)": [mae_rl, mae_rf],
    "RMSE (€)": [rmse_rl, rmse_rf],
    "R²": [r2_rl, r2_rf]
})

display(resultados_modelos)
```

## CÓDIGO A3.18

```python
modelo_xgboost = Pipeline(
    steps=[
        ("preprocesamiento", preprocesador),
        ("modelo", XGBRegressor(
            n_estimators=500,
            learning_rate=0.05,
            max_depth=6,
            subsample=0.8,
            colsample_bytree=0.8,
            random_state=42,
            n_jobs=-1
        ))
    ]
)
```

## CÓDIGO A3.19

```python
modelo_xgboost.fit(X_train, y_train)

print("Modelo XGBoost entrenado correctamente.")
```

## CÓDIGO A3.20

```python
y_pred_xgboost = modelo_xgboost.predict(X_test)

print("Predicciones realizadas correctamente.")
```

## CÓDIGO A3.21

```python
mae_xgb = mean_absolute_error(y_test, y_pred_xgboost)
rmse_xgb = np.sqrt(mean_squared_error(y_test, y_pred_xgboost))
r2_xgb = r2_score(y_test, y_pred_xgboost)

print("Resultados de XGBoost")
print(f"MAE:  {mae_xgb:,.2f} €")
print(f"RMSE: {rmse_xgb:,.2f} €")
print(f"R²:   {r2_xgb:.4f}")
```

## CÓDIGO A3.22

```python
resultados_modelos = pd.DataFrame({
    "Modelo": ["Regresión Lineal", "Random Forest", "XGBoost"],
    "MAE (€)": [mae_rl, mae_rf, mae_xgb],
    "RMSE (€)": [rmse_rl, rmse_rf, rmse_xgb],
    "R²": [r2_rl, r2_rf, r2_xgb]
})

display(resultados_modelos)
```

## CÓDIGO A3.23

```python
cv_repetida = RepeatedKFold(
    n_splits=5,
    n_repeats=20,
    random_state=42
)

print("Validación cruzada repetida configurada correctamente.")
print("Número total de evaluaciones:", 5 * 20)
```

## CÓDIGO A3.24

## CÓDIGO A3.25

```python
scores_rf_cv = cross_val_score(
    modelo_random_forest,
    X,
    y,
    cv=cv_repetida,
    scoring="r2",
    n_jobs=-1
)

print(f"R² medio: {scores_rf_cv.mean():.4f}")
print(f"Desviación estándar: {scores_rf_cv.std():.4f}")
```

## CÓDIGO A3.26

```python
scores_xgb_cv = cross_val_score(
    modelo_xgboost,
    X,
    y,
    cv=cv_repetida,
    scoring="r2",
    n_jobs=-1
)

print(f"R² medio: {scores_xgb_cv.mean():.4f}")
print(f"Desviación estándar: {scores_xgb_cv.std():.4f}")
```

## CÓDIGO A3.28

```python
variables_modelo_a = [
    "size", "rooms", "bathrooms", "floor", "hasLift", "status",
    "propertyType", "typology", "exterior", "newDevelopment"
]

variables_modelo_b = variables_modelo_a + [
    "latitude", "longitude", "dist_metro", "dist_bus", "dist_school",
    "dist_hospital", "dist_health_center", "dist_supermarket",
    "dist_park", "dist_sports"
]
```

## CÓDIGO A3.29

```python
X_a = df[variables_modelo_a].copy()
X_b = df[variables_modelo_b].copy()

X_train_a, X_test_a, y_train_a, y_test_a = train_test_split(
    X_a, y, test_size=0.20, random_state=42
)

X_train_b, X_test_b, y_train_b, y_test_b = train_test_split(
    X_b, y, test_size=0.20, random_state=42
)
```

## CÓDIGO A3.30

```python
variables_categoricas_a = [
    "floor", "hasLift", "status", "propertyType", "typology", "exterior"
]

variables_numericas_a = [
    "size", "rooms", "bathrooms", "newDevelopment"
]

variables_categoricas_b = variables_categoricas_a.copy()

variables_numericas_b = variables_numericas_a + [
    "latitude", "longitude", "dist_metro", "dist_bus", "dist_school",
    "dist_hospital", "dist_health_center", "dist_supermarket",
    "dist_park", "dist_sports"
]

preprocesador_a = ColumnTransformer([
    ("categoricas",
     OneHotEncoder(handle_unknown="ignore", sparse_output=False),
     variables_categoricas_a)
], remainder="passthrough")

preprocesador_b = ColumnTransformer([
    ("categoricas",
     OneHotEncoder(handle_unknown="ignore", sparse_output=False),
     variables_categoricas_b)
], remainder="passthrough")
```

## CÓDIGO A3.31

```python
modelo_rf_a = Pipeline([
    ("preprocesamiento", preprocesador_a),
    ("modelo", RandomForestRegressor(
        n_estimators=500,
        random_state=42,
        n_jobs=-1
    ))
])

modelo_rf_b = Pipeline([
    ("preprocesamiento", preprocesador_b),
    ("modelo", RandomForestRegressor(
        n_estimators=500,
        random_state=42,
        n_jobs=-1
    ))
])
```

## CÓDIGO A3.32

```python
modelo_rf_a.fit(X_train_a, y_train_a)
modelo_rf_b.fit(X_train_b, y_train_b)

print("Modelos Random Forest A y B entrenados correctamente.")
```

## CÓDIGO A3.33

```python
y_pred_rf_a = modelo_rf_a.predict(X_test_a)
y_pred_rf_b = modelo_rf_b.predict(X_test_b)
```

## CÓDIGO A3.34

```python
resultados_rf_a_b = pd.DataFrame({
    "Modelo": ["Random Forest A", "Random Forest B"],
    "MAE (€)": [
        mean_absolute_error(y_test_a, y_pred_rf_a),
        mean_absolute_error(y_test_b, y_pred_rf_b)
    ],
    "RMSE (€)": [
        np.sqrt(mean_squared_error(y_test_a, y_pred_rf_a)),
        np.sqrt(mean_squared_error(y_test_b, y_pred_rf_b))
    ],
    "R²": [
        r2_score(y_test_a, y_pred_rf_a),
        r2_score(y_test_b, y_pred_rf_b)
    ]
})

display(resultados_rf_a_b)
```

## CÓDIGO A3.35

```python
from sklearn.cluster import KMeans
from sklearn.model_selection import GroupKFold

coordenadas = df[["latitude", "longitude"]].copy()

kmeans_espacial = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)

grupos_espaciales = kmeans_espacial.fit_predict(coordenadas)

print(pd.Series(grupos_espaciales).value_counts().sort_index())
```

## CÓDIGO A3.50

```python
fig, ax = plt.subplots(figsize=(8, 7))

for grupo in sorted(np.unique(grupos_espaciales)):
    seleccion = grupos_espaciales == grupo
    ax.scatter(
        df.loc[seleccion, "longitude"],
        df.loc[seleccion, "latitude"],
        s=20,
        label=f"Grupo {grupo}"
    )

ax.set_xlabel("Longitud")
ax.set_ylabel("Latitud")
ax.set_title("Distribución espacial de los grupos")
ax.legend()

plt.tight_layout()
plt.show()
```

## CÓDIGO A3.51

```python
cv_espacial = GroupKFold(n_splits=5)

print("Validación cruzada espacial configurada correctamente.")
```

## CÓDIGO A3.52

```python
scores_xgb_a_espacial = cross_val_score(
    modelo_xgb_a,
    X_a,
    y,
    cv=cv_espacial,
    groups=grupos_espaciales,
    scoring="r2",
    n_jobs=-1
)

print(f"R² medio: {scores_xgb_a_espacial.mean():.4f}")
print(f"Desviación estándar: {scores_xgb_a_espacial.std():.4f}")
```

## CÓDIGO A3.53

```python
scores_xgb_b_espacial = cross_val_score(
    modelo_xgb_b,
    X_b,
    y,
    cv=cv_espacial,
    groups=grupos_espaciales,
    scoring="r2",
    n_jobs=-1
)

print(f"R² medio: {scores_xgb_b_espacial.mean():.4f}")
print(f"Desviación estándar: {scores_xgb_b_espacial.std():.4f}")
```

## CÓDIGO A3.54

```python
resultados_grupos_espaciales = pd.DataFrame({
    "Grupo espacial": range(5),
    "R² - XGBoost A": scores_xgb_a_espacial,
    "R² - XGBoost B": scores_xgb_b_espacial
})

display(resultados_grupos_espaciales)
```
